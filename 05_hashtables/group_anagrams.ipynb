```python
from typing import List 
```

```python
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        if not strs :
            return []
        if len(strs) == 1 :
            return [strs]
        output = {}
        for i in strs :
            key = "".join(sorted(i))
            output[key] = output.get(key , [] ) + [i]
        return list(output.values())

        # O(k log k)
```

```python
def norm(groups):
    return sorted(sorted(g) for g in groups)

s = Solution()
assert norm(s.groupAnagrams(["eat", "tea", "tan", "ate", "nat", "bat"])) == \
       norm([["eat", "tea", "ate"], ["tan", "nat"], ["bat"]])          # مثال المسألة
assert s.groupAnagrams([]) == []                                       # فاضية
assert s.groupAnagrams(["a"]) == [["a"]]                               # كلمة واحدة
assert s.groupAnagrams([""]) == [[""]]                                 # كلمة فاضية
assert norm(s.groupAnagrams(["abc", "def", "ghi"])) == \
       norm([["abc"], ["def"], ["ghi"]])                               # مفيش anagrams خالص
assert norm(s.groupAnagrams(["ab", "ba", "ab"])) == \
       norm([["ab", "ba", "ab"]])                                      # كلمات متكررة
assert norm(s.groupAnagrams(["aab", "abb"])) == \
       norm([["aab"], ["abb"]])                                        # نفس الحروف بعدد مختلف
print("groupAnagrams passed ✅")
```
