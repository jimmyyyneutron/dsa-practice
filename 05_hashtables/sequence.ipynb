```python
from typing import List
```

```python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        nums_set=set(nums) 
        longest = 0
        for i in nums_set :
            if not i -1 in nums_set:
                current = i
                sequence = 1
                while current + 1 in nums_set:
                    sequence += 1
                    current += 1
                longest = max(longest, sequence)
        return longest
```

```python
s = Solution()
assert s.longestConsecutive([100, 4, 200, 1, 3, 2]) == 4          # مثال المسألة
assert s.longestConsecutive([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]) == 9  # تكرار + سلسلة طويلة
assert s.longestConsecutive([]) == 0                              # فاضية
assert s.longestConsecutive([5]) == 1                             # رقم واحد
assert s.longestConsecutive([1, 2, 2, 3]) == 3                    # أرقام متكررة
assert s.longestConsecutive([10, 30, 20]) == 1                    # مفيش أي رقمين ورا بعض
assert s.longestConsecutive([-2, -1, 0, 1]) == 4                  # أرقام سالبة
assert s.longestConsecutive(list(range(100000, 0, -1))) == 100000 # list كبيرة بالعكس
print("longestConsecutive passed ✅")
```
