```python
from typing import List
```

```python
class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        dic = {}
        for i in nums :
            dic[i] = dic.get(i , 0) + 1
        return sorted(dic , key=dic.get , reverse=True)[:k]
  # O(nlogn)
```

```python
s = Solution()
assert sorted(s.topKFrequent([1, 1, 1, 2, 2, 3], 2)) == [1, 2]       # مثال المسألة
assert sorted(s.topKFrequent([1], 1)) == [1]                         # عنصر واحد
assert sorted(s.topKFrequent([4, 4, 5, 5, 5, 6], 1)) == [5]          # k = 1
assert sorted(s.topKFrequent([1, 2, 3], 3)) == [1, 2, 3]             # k = عدد الأرقام المختلفة
assert sorted(s.topKFrequent([-1, -1, 2, 2, 2, 0], 2)) == [-1, 2]    # أرقام سالبة
assert sorted(s.topKFrequent([7, 7, 8, 8, 8, 9], 2)) == [7, 8]       # الأكتر مش في أول الـ list
print("topKFrequent passed ✅")
```
