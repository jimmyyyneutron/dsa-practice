```python
from typing import List
```

```python
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        m = len(matrix)
        n= len(matrix[0])
        d = m * n 
        low = 0
        high = d-1 
        while low <= high :
            mid = (low + high ) // 2 
            row = mid // n
            col = mid % n
            value = matrix[row][col]
            if value == target :
                return True
            elif value < target :
                low = mid+1 
            else :
                high = mid-1 
        return False
```

```python
if __name__ == "__main__":
    s = Solution()
    matrix = [[1, 2, 4, 8], [10, 11, 12, 13], [14, 20, 30, 40]]

    tests = [
        (matrix, 10, True),       # مثال المسألة
        (matrix, 15, False),      # مش موجود
        (matrix, 1, True),        # أول عنصر
        (matrix, 40, True),       # آخر عنصر
        (matrix, 0, False),       # أصغر من الكل
        (matrix, 99, False),      # أكبر من الكل
        ([[5]], 5, True),         # عنصر واحد
        ([[1, 3, 5, 7]], 7, True),        # صف واحد
        ([[1], [3], [5], [7]], 4, False), # عمود واحد
    ]

    for m, target, expected in tests:
        result = s.searchMatrix(m, target)
        print("✓" if result == expected else "✗", target, "→", result)
```
