```python
import math
```

```python
class Solution:
    def minEatingSpeed(self, piles: List[int], h: int) -> int:
        low = 1
        high = max(piles)
        k = max(piles)
        while low <= high :
            mid = (low + high) // 2
            hours = 0 
            for i in piles:
                hours += math.ceil(i/mid) 
            if hours <= h :
                high = mid -1 
                k = mid 
            else:
                low = mid +1 
        return k
        
```

```python
s = Solution()
tests = [
    ([1, 4, 3, 2], 9, 2),              # Example 1
    ([25, 10, 23, 4], 4, 25),          # Example 2: ساعة لكل كومة
    ([3, 6, 7, 11], 8, 4),
    ([30, 11, 23, 4, 20], 5, 30),
    ([30, 11, 23, 4, 20], 6, 23),
    ([30], 5, 6),                      # المثال اللي لخبطك
    ([5], 1, 5),                       # كومة واحدة وساعة واحدة
    ([1, 1, 1, 1], 4, 1),              # أقل سرعة ممكنة
    ([1000000000], 2, 500000000),      # أرقام كبيرة
]

for piles, h, expected in tests:
    result = s.minEatingSpeed(piles, h)
    print("✓" if result == expected else "✗", piles, "h =", h, "→", result)
```
