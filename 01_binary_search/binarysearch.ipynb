```python
def binary_search(arr , num):
    less = 0
    max = len(arr) - 1
    while less <= max :
        mid = (less + max) // 2
        guess = arr[mid]
        if guess == num :
            return mid
        elif guess < num :
        # too low
            less = mid + 1  
        else:
        # too high
            max = mid- 1     
    return None     
```

```python
print(main([1, 3, 5, 7, 9], 7))   # expect 3
print(main([1, 3, 5, 7, 9], 1))   # expect 0
print(main([1, 3, 5, 7, 9], 4))   # expect None
```
