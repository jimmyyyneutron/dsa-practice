```python
def binary_search(arr , num):
    less = 0
    max = len(arr) - 1
    while less <= max :
        mid = (less + max) // 2
        guess = arr[mid]
        if guess == num :
            return mid
        elif guess < num :
        # too low
            less = mid + 1  
        else:
        # too high
            max = mid- 1     
    return None     
```

```python
print(binary_search([1, 3, 5, 7, 9], 7))   # expect 3
print(binary_search([1, 3, 5, 7, 9], 1))   # expect 0
print(binary_search([1, 3, 5, 7, 9], 4))   # expect None
```

```python
def binary_search(arr , num , max=None , less=0):
    if max == None:
        max = len(arr) - 1 
    if less > max :
        return None
    mid = (less + max) // 2
    guess = arr[mid]
    if guess == num :
        return mid
    elif guess < num :
    # too low
        return binary_search(arr, num, max, mid + 1)  
    else:
    # too high
        return binary_search(arr, num, mid - 1, less) 
  
# recursion version
```

```python
assert binary_search([1, 3, 5, 7, 9], 7) == 3       # في النص
assert binary_search([1, 3, 5, 7, 9], 1) == 0       # أول عنصر
assert binary_search([1, 3, 5, 7, 9], 9) == 4       # آخر عنصر
assert binary_search([1, 3, 5, 7, 9], 4) is None    # مش موجود
assert binary_search([], 5) is None                 # list فاضية
assert binary_search([5], 5) == 0                   # عنصر واحد
assert binary_search(list(range(100000)), 99999) == 99999   # list كبيرة
print("binary_search passed ✅")
```
