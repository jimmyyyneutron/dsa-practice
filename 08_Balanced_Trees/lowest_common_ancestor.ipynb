```python
from typing import Optional, List
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
# بتبني شجرة من list بنفس طريقة LeetCode (طبقة طبقة، و None يعني مفيش node)
def build_tree(values):
    if not values or values[0] is None:
        return None
    root = TreeNode(values[0])
    queue = deque([root])
    i = 1
    while queue and i < len(values):
        node = queue.popleft()
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    return root
```

```python
# بتحوّل الشجرة لـ list تاني، عشان نقارن بسهولة
def to_list(root):
    result = []
    queue = deque([root])
    while queue:
        node = queue.popleft()
        if node:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            result.append(None)
    while result and result[-1] is None:
        result.pop()
    return result
```

```python
# بتحوّل الشجرة لـ list تاني، عشان نقارن بسهولة
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def lowestCommonAncestor(self, root: TreeNode, p: TreeNode, q: TreeNode) -> TreeNode:
        if p.val<root.val and q.val < root.val :
            return self.lowestCommonAncestor(root.left , p , q)
        elif p.val>root.val and q.val > root.val :
            return self.lowestCommonAncestor(root.right , p , q)
        else :
            return root
```

```python
def find(root, val):
    while root and root.val != val:
        root = root.left if val < root.val else root.right
    return root

s = Solution()
t = build_tree([5, 3, 8, 1, 4, 7, 9])
assert s.lowestCommonAncestor(t, find(t, 1), find(t, 4)).val == 3    # نفس الفرع
assert s.lowestCommonAncestor(t, find(t, 1), find(t, 9)).val == 5    # فرعين مختلفين
assert s.lowestCommonAncestor(t, find(t, 7), find(t, 9)).val == 8    # المثال اللي مشيناه
assert s.lowestCommonAncestor(t, find(t, 3), find(t, 4)).val == 3    # واحد فيهم هو الأب
assert s.lowestCommonAncestor(t, find(t, 5), find(t, 9)).val == 5    # واحد فيهم هو الـ root
assert s.lowestCommonAncestor(t, find(t, 9), find(t, 7)).val == 8    # p و q متبدلين
print("lowestCommonAncestor passed ✅")
```
