```python
from typing import Optional, List
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
# بتبني شجرة من list بنفس طريقة LeetCode (طبقة طبقة، و None يعني مفيش node)
def build_tree(values):
    if not values or values[0] is None:
        return None
    root = TreeNode(values[0])
    queue = deque([root])
    i = 1
    while queue and i < len(values):
        node = queue.popleft()
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    return root
```

```python
# بتحوّل الشجرة لـ list تاني، عشان نقارن بسهولة
def to_list(root):
    result = []
    queue = deque([root])
    while queue:
        node = queue.popleft()
        if node:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            result.append(None)
    while result and result[-1] is None:
        result.pop()
    return result
```

```python
class Solution:
    def invertTree(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        if not root :
            return None
        root.left , root.right = root.right , root.left
        self.invertTree(root.left)
        self.invertTree(root.right)
        return root
# Complexity: O(n) time, O(h) space

```

```python
s = Solution()
assert to_list(s.invertTree(build_tree([4, 2, 7, 1, 3, 6, 9]))) == [4, 7, 2, 9, 6, 3, 1]   # مثال المسألة
assert to_list(s.invertTree(build_tree([2, 1, 3]))) == [2, 3, 1]                            # 3 nodes
assert to_list(s.invertTree(build_tree([]))) == []                                           # فاضية
assert to_list(s.invertTree(build_tree([1]))) == [1]                                         # node واحدة
assert to_list(s.invertTree(build_tree([1, 2]))) == [1, None, 2]                             # ابن شمال بس، فيبقى يمين
print("invertTree passed ✅")
```
