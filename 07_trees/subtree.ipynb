```python
from typing import Optional, List
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
# بتبني شجرة من list بنفس طريقة LeetCode (طبقة طبقة، و None يعني مفيش node)
def build_tree(values):
    if not values or values[0] is None:
        return None
    root = TreeNode(values[0])
    queue = deque([root])
    i = 1
    while queue and i < len(values):
        node = queue.popleft()
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    return root
```

```python
# بتحوّل الشجرة لـ list تاني، عشان نقارن بسهولة
def to_list(root):
    result = []
    queue = deque([root])
    while queue:
        node = queue.popleft()
        if node:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            result.append(None)
    while result and result[-1] is None:
        result.pop()
    return result
```

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:  
    
    
    def isSameTree(self, p: Optional[TreeNode], q: Optional[TreeNode]) -> bool:
        if p == None and q == None :
            return True
        elif p == None or q == None :
            return False
        elif p.val != q.val :
            return False 

        return self.isSameTree(p.left , q.left ) and self.isSameTree(p.right , q.right ) 
    
    
    def isSubtree(self, root: Optional[TreeNode], subRoot: Optional[TreeNode]) -> bool:
        if root is None :
            return False
        elif self.isSameTree(root , subRoot) :
            return True
        
        return self.isSubtree(root.left, subRoot ) or self.isSubtree(root.right ,subRoot)
```

```python
s = Solution()
assert s.isSubtree(build_tree([3, 4, 5, 1, 2]), build_tree([4, 1, 2])) == True                        # موجودة
assert s.isSubtree(build_tree([3, 4, 5, 1, 2, None, None, None, None, 0]), build_tree([4, 1, 2])) == False  # شبهها بس فيه node زيادة تحت
assert s.isSubtree(build_tree([1, 2, 3]), build_tree([1, 2, 3])) == True                              # الشجرة كلها
assert s.isSubtree(build_tree([1, 2, 3]), build_tree([3])) == True                                    # ورقة
assert s.isSubtree(build_tree([1, 2, 3]), build_tree([4])) == False                                   # مش موجودة
assert s.isSubtree(build_tree([1, 1]), build_tree([1])) == True                                       # قيم متكررة
print("isSubtree passed ✅")
```
