```python
class Solution1:
    def twoSum(self, nums: List[int], target: int) -> List[int]:

        for i in range(len(nums)) :
            for k in range (i+1,len(nums)):
                if nums[i] + nums[k] == target :
                    return [i,k]
```

```python
for name, s in [("Solution1", Solution1()), ("Solution2", Solution2())]:
    assert s.twoSum([2, 7, 11, 15], 9) == [0, 1]       # الإجابة في الأول
    assert s.twoSum([3, 2, 4], 6) == [1, 2]            # الإجابة في الآخر
    assert s.twoSum([3, 3], 6) == [0, 1]               # نفس الرقم مرتين
    assert s.twoSum([2, 5, 5, 11], 10) == [1, 2]       # ميجمعش العنصر مع نفسه
    assert s.twoSum([-3, 4, 3, 90], 0) == [0, 2]       # أرقام سالبة
    assert s.twoSum([0, 4, 3, 0], 0) == [0, 3]         # أصفار
    big = list(range(5000))                             # list كبيرة، والإجابة في آخرها
    assert s.twoSum(big, 4998 + 4999) == [4998, 4999]
    print(name, "passed ✅")
```

```python
class Solution2:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {}
        for i in range(len(nums)):
            diff = target - nums[i]
            if diff in seen :
                return [seen[diff] , i]
            seen[nums[i]] = i
```

```python
for name, s in [("Solution1", Solution1()), ("Solution2", Solution2())]:
    assert s.twoSum([2, 7, 11, 15], 9) == [0, 1]       # الإجابة في الأول
    assert s.twoSum([3, 2, 4], 6) == [1, 2]            # الإجابة في الآخر
    assert s.twoSum([3, 3], 6) == [0, 1]               # نفس الرقم مرتين
    assert s.twoSum([2, 5, 5, 11], 10) == [1, 2]       # ميجمعش العنصر مع نفسه
    assert s.twoSum([-3, 4, 3, 90], 0) == [0, 2]       # أرقام سالبة
    assert s.twoSum([0, 4, 3, 0], 0) == [0, 3]         # أصفار
    big = list(range(5000))                             # list كبيرة، والإجابة في آخرها
    assert s.twoSum(big, 4998 + 4999) == [4998, 4999]
    print(name, "passed ✅")
```
