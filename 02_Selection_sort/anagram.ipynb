```python
class Solution1:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t) :
            return False
        check = set()
        for i in range (len(s)) :
            found = False
            for k in range(len(t)):
                if s[i] == t[k] and k not in check : 
                    found = True
                    check.add(k)
                    break
            if not found :
                return False
        return True 
# complexity O(n^2)
```

```python
s1 = Solution1()
assert s1.isAnagram("anagram", "nagaram") == True    # anagram عادي
assert s1.isAnagram("aab", "abb") == False           # نفس الحروف بعدد مختلف
assert s1.isAnagram("rat", "rats") == False          # طول مختلف
print("Solution1 passed ✅")
```

```python
class Solution2:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t) :
            return False
        return sorted(s) == sorted(t)
# complexity O(nlogn)
```

```python
s2 = Solution2()
assert s2.isAnagram("listen", "silent") == True      # anagram عادي
assert s2.isAnagram("rat", "car") == False           # حروف مختلفة
assert s2.isAnagram("a", "a") == True                # حرف واحد
print("Solution2 passed ✅")
```

```python
class Solution3:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t) :
            return False
        sum_s = {}
        sum_t = {}
        for i in range(len(s)):
            sum_s[s[i]] = sum_s.get(s[i] , 0) + 1 
            sum_t[t[i]] = sum_t.get(t[i] , 0) + 1
        return sum_s == sum_t
    # complexity O(n)
```

```python
s3 = Solution3()
assert s3.isAnagram("aacc", "ccac") == False         # نفس الحروف بعدد مختلف
assert s3.isAnagram("", "") == True                  # كلمتين فاضيين
assert s3.isAnagram("abc" * 10000, "cba" * 10000) == True   # كلمات طويلة
print("Solution3 passed ✅")
```
