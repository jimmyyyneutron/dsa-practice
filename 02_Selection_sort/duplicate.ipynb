```python
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        num = set()
        for n in nums :
            if n in num :
                return True
            else :
                num.add(n)
        return False
```

```python
s = Solution()
assert s.hasDuplicate([1, 2, 3, 1]) == True        # تكرار عادي
assert s.hasDuplicate([1, 2, 3, 4]) == False       # مفيش تكرار
assert s.hasDuplicate([]) == False                 # list فاضية
assert s.hasDuplicate([7]) == False                # عنصر واحد
assert s.hasDuplicate([5, 5]) == True              # عنصرين متكررين بس
assert s.hasDuplicate([-1, -2, -1]) == True        # أرقام سالبة
assert s.hasDuplicate([0, 0]) == True              # أصفار
assert s.hasDuplicate([1, 2, 3, 4, 5, 1]) == True  # التكرار في الآخر
assert s.hasDuplicate(list(range(100000))) == False             # list كبيرة من غير تكرار
assert s.hasDuplicate(list(range(100000)) + [99999]) == True    # list كبيرة والتكرار في آخرها

print("All tests passed ✅")
```

```python
from typing import List
```

```python
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        nums.sort()
        for i in range(1 , len(nums)):
            if nums[i] == nums[i-1]:
                return True
        return False
```

```python
s = Solution()
assert s.hasDuplicate([1, 2, 3, 1]) == True        # تكرار عادي
assert s.hasDuplicate([1, 2, 3, 4]) == False       # مفيش تكرار
assert s.hasDuplicate([]) == False                 # list فاضية
assert s.hasDuplicate([7]) == False                # عنصر واحد
assert s.hasDuplicate([5, 5]) == True              # عنصرين متكررين بس
assert s.hasDuplicate([-1, -2, -1]) == True        # أرقام سالبة
assert s.hasDuplicate([0, 0]) == True              # أصفار
assert s.hasDuplicate([3, 1, 4, 1]) == True        # التكرار مش جنب بعض قبل الترتيب
assert s.hasDuplicate([9, 8, 7, 6]) == False       # مترتبة بالعكس
assert s.hasDuplicate(list(range(100000))) == False             # list كبيرة من غير تكرار
assert s.hasDuplicate(list(range(100000)) + [99999]) == True    # list كبيرة والتكرار في آخرها

# اختبار إضافي: الحل ده بيغيّر الـ list الأصلية
arr = [3, 1, 2]
s.hasDuplicate(arr)
print(arr)  # هتطلع [1, 2, 3] مش [3, 1, 2]

print("All tests passed ✅")
```
