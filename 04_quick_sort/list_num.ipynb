```python
def list_count(arr):
    if arr == []:
        return 0 
    return 1 + list_count(arr[1:])
```

```python
assert list_count([2, 4, 6]) == 3         # عادي
assert list_count([]) == 0                # فاضية
assert list_count([5]) == 1               # عنصر واحد
assert list_count([0, 0, 0]) == 3         # أصفار بتتعد برضه
assert list_count(["a", "b"]) == 2        # strings مش أرقام بس
assert list_count(list(range(500))) == 500    # list طويلة
print("list_count passed ✅")
```
