```python
def quick_sort(arr):
    if len(arr) < 2:
        return arr
    pivot = arr[0]
    lower = [i for i in arr[1:] if i <= pivot]
    higher = [i for i in arr[1:] if i > pivot]
    return quick_sort(lower) + [pivot] + quick_sort(higher)
```

```python
print(quick_sort([10,5,2,3]))
```

```python
assert quick_sort([10, 5, 2, 3]) == [2, 3, 5, 10]     # عادي
assert quick_sort([]) == []                           # فاضية
assert quick_sort([7]) == [7]                         # عنصر واحد
assert quick_sort([3, 1, 3, 2]) == [1, 2, 3, 3]       # أرقام متكررة
assert quick_sort([5, 5, 5]) == [5, 5, 5]             # كلها زي بعض
assert quick_sort([1, 2, 3, 4]) == [1, 2, 3, 4]       # مترتبة أصلًا
assert quick_sort([4, 3, 2, 1]) == [1, 2, 3, 4]       # مترتبة بالعكس
assert quick_sort([-2, 0, -5, 3]) == [-5, -2, 0, 3]   # أرقام سالبة

import random
nums = [random.randint(-1000, 1000) for _ in range(1000)]
assert quick_sort(nums) == sorted(nums)               # 1000 رقم عشوائي
print("quick_sort passed ✅")
```
