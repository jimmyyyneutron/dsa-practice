```python
from typing import List
```

```python
# Definition for a pair.
class Pair:
    def __init__(self, key: int, value: str):
        self.key = key
        self.value = value
```

```python
class Solution:
    def insertionSort(self, pairs: List[Pair]) -> List[List[Pair]]:
        if len(pairs) == 0 :
            return []
        states =[]
        copiedArr = states
        for i in range(len(pairs)) :
            while i > 0 and pairs[i].key < pairs[i-1].key :
                pairs[i] , pairs[i -1] = pairs[i-1] , pairs[i]
                i -= 1
            states.append(list(pairs))
        return states
# Complexity: O(n²) time
```

```python
def run(data):
    pairs = [Pair(k, v) for k, v in data]
    result = Solution().insertionSort(pairs)
    return [[(p.key, p.value) for p in state] for state in result]
```

```python
assert run([(5, "apple"), (2, "banana"), (9, "cherry")]) == [
    [(5, "apple"), (2, "banana"), (9, "cherry")],
    [(2, "banana"), (5, "apple"), (9, "cherry")],
    [(2, "banana"), (5, "apple"), (9, "cherry")],
]

# فاضية
assert run([]) == []

# عنصر واحد
assert run([(1, "a")]) == [[(1, "a")]]

# مترتبة بالعكس
assert run([(3, "c"), (2, "b"), (1, "a")]) == [
    [(3, "c"), (2, "b"), (1, "a")],
    [(2, "b"), (3, "c"), (1, "a")],
    [(1, "a"), (2, "b"), (3, "c")],
]

# نفس الـ key: الترتيب الأصلي لازم يفضل زي ما هو
assert run([(3, "cat"), (3, "bird"), (2, "dog")]) == [
    [(3, "cat"), (3, "bird"), (2, "dog")],
    [(3, "cat"), (3, "bird"), (2, "dog")],
    [(2, "dog"), (3, "cat"), (3, "bird")],
]

# عدد الحالات = عدد العناصر
assert len(run([(i, "x") for i in range(10, 0, -1)])) == 10

print("insertionSort passed ✅")
```
