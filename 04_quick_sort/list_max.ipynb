```python
def list_max(arr):
    if len(arr) == 0 :
        return None
    if len(arr) == 1 : 
        return arr[0]
    else :
        max_num = list_max(arr[1:])
        if arr[0] > max_num :
            return arr[0]
        else:
            return max_num 
    
```

```python
assert list_max([3, 7, 2]) == 7       # الأكبر في النص
assert list_max([9, 1, 4]) == 9       # الأكبر في الأول
assert list_max([1, 4, 9]) == 9       # الأكبر في الآخر
assert list_max([5]) == 5             # عنصر واحد
assert list_max([]) is None           # فاضية
assert list_max([-5, -2, -9]) == -2   # كلها سالبة
assert list_max([4, 4, 4]) == 4       # متساويين
print("list_max passed ✅")
```
