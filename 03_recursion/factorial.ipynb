```python
def fact(x) :
    if x < 0:
        raise ValueError("factorial not valid for negative numbers")
    if x <= 1:
        return 1
    else :
        return x * fact(x -1)
```

```python
assert fact(0) == 1
assert fact(1) == 1
assert fact(5) == 120
assert fact(10) == 3628800
print("All tests passed ✅")
```
