```python
class MinStack:

    def __init__(self):
        self.stack = []
        self.minstack= []
    def push(self, val: int) -> None:
        self.stack.append(val)
        if not self.minstack :
            self.minstack.append(val)
        else: 
            self.minstack.append(min(val , self.minstack[-1]))
    def pop(self) -> None:
        self.stack.pop()
        self.minstack.pop()
    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        return self.minstack[-1]
        
```

```python
m = MinStack()
m.push(5)
assert m.getMin() == 5 and m.top() == 5     # عنصر واحد
m.push(3)
assert m.getMin() == 3                      # رقم أصغر دخل
m.push(7)
assert m.getMin() == 3 and m.top() == 7     # رقم أكبر مش بيغيّر الـ min
m.pop()
assert m.getMin() == 3 and m.top() == 3     # شلنا الـ 7
m.pop()
assert m.getMin() == 5                      # شلنا الأصغر، فرجع القديم

m2 = MinStack()
m2.push(2); m2.push(2); m2.push(1)
m2.pop()
assert m2.getMin() == 2                     # أرقام متكررة

m3 = MinStack()
m3.push(-2); m3.push(0); m3.push(-3)
assert m3.getMin() == -3                    # أرقام سالبة
m3.pop()
assert m3.top() == 0 and m3.getMin() == -2
print("MinStack passed ✅")
```
