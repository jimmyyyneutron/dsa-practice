```python
class Solution:
    def isValid(self, s: str) -> bool:
        pairs = { "(" : ")" , "[": "]" , "{" : "}" }
        stack = []
        for i in s : 
            if i in pairs:
                stack.append(i)
            else :
                if not stack :
                    return False
                if pairs[stack[-1]] != i :
                    return False
                stack.pop()
        return len(stack) == 0
```

```python
s = Solution()
assert s.isValid("()") == True              # زوج واحد
assert s.isValid("()[]{}") == True          # أنواع ورا بعض
assert s.isValid("{[()]}") == True          # متداخلين
assert s.isValid("(]") == False             # نوع غلط
assert s.isValid("([)]") == False           # ترتيب غلط
assert s.isValid("((") == False             # اتفتح ومااتقفلش
assert s.isValid(")") == False              # قفل من غير فتح
assert s.isValid("())") == False            # قفل زيادة في الآخر
assert s.isValid("") == True                # string فاضي
print("isValid passed ✅")
```
