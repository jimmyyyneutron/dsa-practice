```python
class Solution:
    def evalRPN(self, tokens: List[str]) -> int:
        stack = []
        for i in tokens :
            if i in ["+","*","/","-"] :
                b = stack.pop()
                a = stack.pop()
                if i == "+" :
                    stack.append(a + b)
                elif i == "-" :
                    stack.append(a -  b)
                elif i == "*" :
                    stack.append(a * b)
                else :
                    stack.append(int(a/b))
            else :
                stack.append(int(i))
        return stack[0]
```

```python
s = Solution()
assert s.evalRPN(["2", "1", "+", "3", "*"]) == 9          # (2+1)*3
assert s.evalRPN(["4", "13", "5", "/", "+"]) == 6         # 4+(13/5)
assert s.evalRPN(["5"]) == 5                              # رقم لوحده
assert s.evalRPN(["5", "3", "-"]) == 2                    # ترتيب الطرح: 5-3 مش 3-5
assert s.evalRPN(["8", "2", "/"]) == 4                    # ترتيب القسمة: 8/2 مش 2/8
assert s.evalRPN(["-6", "4", "/"]) == -1                  # تقريب ناحية الصفر مش -2
assert s.evalRPN(["7", "-3", "/"]) == -2                  # رقم سالب في المقام
assert s.evalRPN(["-11", "3", "*"]) == -33                # رقم سالب مش علامة طرح
assert s.evalRPN(["10", "6", "9", "3", "+", "-11", "*",
                  "/", "*", "17", "+", "5", "+"]) == 22   # المثال الطويل من LeetCode
print("evalRPN passed ✅")
```
