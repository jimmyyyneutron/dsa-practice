# Coin Change (LeetCode 322)

```python
import heapq
import random
from typing import List
```

```python
class Solution:
    def coinChange(self, coins: List[int], amount: int) -> int:
        grid = [float("inf")] * (amount + 1)
        grid[0] = 0
        for i in range(1, amount + 1):
            for k in coins :
                if k > i :
                    continue
                grid[i] = min(grid[i-k] + 1, grid[i])
        if grid[-1] != float("inf") :
            return grid[-1]
        else :
            return -1
```

## الاختبارات

```python
s = Solution()

# 1. أمثلة LeetCode
print(s.coinChange([1, 2, 5], 11) == 3)
print(s.coinChange([2], 3) == -1)
print(s.coinChange([1], 0) == 0)

# 2. المثال اللي الـ Greedy بيغلط فيه
print(s.coinChange([1, 3, 4], 6) == 2)

# 3. الجدول اللي ملّيناه بالورقة
print([s.coinChange([1, 2, 5], a) for a in range(7)] == [0, 1, 1, 2, 2, 1, 2])

# 4. مستحيل
print(s.coinChange([5, 10], 3) == -1)

# 5. مقارنة بحل BFS
from collections import deque
def slow(coins, amount):
    seen = {0}
    q = deque([(0, 0)])
    while q:
        total, steps = q.popleft()
        if total == amount:
            return steps
        for c in coins:
            nxt = total + c
            if nxt <= amount and nxt not in seen:
                seen.add(nxt)
                q.append((nxt, steps + 1))
    return -1

ok = True
for _ in range(300):
    coins = random.sample(range(1, 20), random.randint(1, 4))
    amount = random.randint(0, 60)
    if s.coinChange(coins, amount) != slow(coins, amount):
        ok = False
print(ok)
```
