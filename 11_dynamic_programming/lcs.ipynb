# Longest Common Subsequence (LeetCode 1143)

```python
import heapq
import random
from typing import List
```

```python
class Solution:
    def longestCommonSubsequence(self, text1: str, text2: str) -> int:
        rows = len(text1) + 1
        cols = len(text2) + 1
        dp = [[0] * cols for _ in range(rows)]
        for i in range(1,rows) :
            for j in range(1,cols):
                if text1[i-1] == text2[j-1]:
                    dp[i][j] = dp[i - 1][j -1 ] + 1
                else :
                    dp[i][j] = max(dp[i-1][j] , dp[i][j-1])
        return dp[-1][-1]
```

## الاختبارات

```python
s = Solution()

# 1. أمثلة LeetCode
print(s.longestCommonSubsequence("abcde", "ace") == 3)
print(s.longestCommonSubsequence("abc", "abc") == 3)
print(s.longestCommonSubsequence("abc", "def") == 0)

# 2. الجدول اللي ملّيناه بالورقة
print(s.longestCommonSubsequence("abc", "ac") == 2)

# 3. حرف واحد
print(s.longestCommonSubsequence("a", "a") == 1)
print(s.longestCommonSubsequence("a", "b") == 0)

# 4. الترتيب مهم
print(s.longestCommonSubsequence("abc", "cba") == 1)

# 5. مقارنة بحل recursion + memo
from functools import lru_cache
def slow(a, b):
    @lru_cache(None)
    def go(i, j):
        if i == len(a) or j == len(b):
            return 0
        if a[i] == b[j]:
            return 1 + go(i + 1, j + 1)
        return max(go(i + 1, j), go(i, j + 1))
    return go(0, 0)

ok = True
for _ in range(300):
    a = "".join(random.choice("abcd") for _ in range(random.randint(1, 12)))
    b = "".join(random.choice("abcd") for _ in range(random.randint(1, 12)))
    if s.longestCommonSubsequence(a, b) != slow(a, b):
        ok = False
print(ok)
```
