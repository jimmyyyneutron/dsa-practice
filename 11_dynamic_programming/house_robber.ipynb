# House Robber (LeetCode 198)

```python
import heapq
import random
from typing import List
```

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        if len(nums) == 1 :
            return nums[0]
        if len(nums) == 2 :
            return max(nums[0],nums[1])
        best = [0] * len(nums)
        best[0] = nums[0]
        best[1] = max(nums[0],nums[1])
        for i in range(2 , len(nums)) :
            best[i] = max(best[i-1] , nums[i] + best[i-2])
        return best[-1]
```

## الاختبارات

```python
s = Solution()

# 1. أمثلة LeetCode
print(s.rob([1, 2, 3, 1]) == 4)
print(s.rob([2, 7, 9, 3, 1]) == 12)

# 2. بيت واحد وبيتين
print(s.rob([5]) == 5)
print(s.rob([2, 9]) == 9)

# 3. أحسن حاجة تسيب بيتين ورا بعض
print(s.rob([2, 1, 1, 2]) == 4)

# 4. أصفار
print(s.rob([0, 0, 0]) == 0)

# 5. مقارنة بالحل البطيء (كل الاختيارات)
def slow(nums):
    best = 0
    n = len(nums)
    for mask in range(1 << n):
        if mask & (mask >> 1):
            continue
        best = max(best, sum(nums[i] for i in range(n) if mask >> i & 1))
    return best

ok = True
for _ in range(300):
    nums = [random.randint(0, 50) for _ in range(random.randint(1, 12))]
    if s.rob(nums) != slow(nums):
        ok = False
print(ok)
```
