# Partition Equal Subset Sum (LeetCode 416)

```python
import heapq
import random
from typing import List
```

```python
class Solution:
    def canPartition(self, nums: List[int]) -> bool:
        if sum(nums) % 2 :
            return False
        dp =set()
        dp.add(0)
        target = sum(nums) // 2
        for i in nums :
            temp = set()
            for k in dp :
                temp.add(k)
                temp.add(k+i)
            dp=temp
        return target in dp
```

## الاختبارات

```python
s = Solution()

# 1. أمثلة LeetCode
print(s.canPartition([1, 5, 11, 5]) == True)
print(s.canPartition([1, 2, 3, 5]) == False)

# 2. المجموع فردي
print(s.canPartition([1, 2, 4]) == False)

# 3. رقمين متساويين
print(s.canPartition([7, 7]) == True)

# 4. رقم واحد
print(s.canPartition([2]) == False)

# 5. مقارنة بالحل البطيء (كل التقسيمات)
def slow(nums):
    total = sum(nums)
    n = len(nums)
    for mask in range(1 << n):
        if 2 * sum(nums[i] for i in range(n) if mask >> i & 1) == total:
            return True
    return False

ok = True
for _ in range(300):
    nums = [random.randint(1, 20) for _ in range(random.randint(1, 12))]
    if s.canPartition(nums) != slow(nums):
        ok = False
print(ok)
```
