# Climbing Stairs (LeetCode 70) - 3 طرق

```python
import heapq
import random
from typing import List
```

## الطريقة 1: Top-down (recursion + memo)

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        steps = {}
        def helper(n):
            if n ==1 :
                return 1
            if n == 2 :
                return 2
            if n in steps :
                return steps[n]
            else:
                steps[n] = helper(n-1) + helper(n-2)
                return steps[n]
        return helper(n)
```

```python
TopDown = Solution
```

## الطريقة 2: Bottom-up (جدول)

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        if n==1 :
            return 1
        ways = [0] * (n+1)
        ways[1] = 1
        ways[2] = 2
        for i in range(3 , n+1 ) :
            ways[i] = ways[i-1] + ways[i-2]
        return ways[n]
```

```python
Table = Solution
```

## الطريقة 3: متغيرين (O(1) space)

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        if n==1 :
            return 1
        prev = 1
        cur = 2
        for i in range(3, n+1):
            new = prev + cur
            prev = cur
            cur = new
        return cur
```

```python
TwoVars = Solution
```

## الاختبارات (على الـ 3 طرق)

```python
expected = {1: 1, 2: 2, 3: 3, 4: 5, 5: 8, 10: 89, 45: 1836311903}

for name, cls in [("TopDown", TopDown), ("Table", Table), ("TwoVars", TwoVars)]:
    s = cls()
    ok = all(s.climbStairs(n) == ans for n, ans in expected.items())
    print(name, ok)

# الـ 3 طرق لازم يطلّعوا نفس النتيجة
print(all(TopDown().climbStairs(n) == Table().climbStairs(n) == TwoVars().climbStairs(n) for n in range(1, 46)))
```
