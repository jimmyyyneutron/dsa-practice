# Jump Game (LeetCode 55)

```python
import heapq
import random
from typing import List
```

```python
class Solution:
    def canJump(self, nums: List[int]) -> bool:
        farthest = 0
        for i in range(len(nums)) :
            if i > farthest :
                return False
            if farthest < (i + nums[i]) :
                farthest = i + nums[i]
        return True
```

## الاختبارات

```python
s = Solution()

# 1. أمثلة LeetCode
print(s.canJump([2, 3, 1, 1, 4]) == True)
print(s.canJump([3, 2, 1, 0, 4]) == False)

# 2. خانة واحدة: انت واقف في الآخر أصلًا
print(s.canJump([0]) == True)

# 3. صفر في الأول
print(s.canJump([0, 1]) == False)

# 4. صفر في الآخر عادي
print(s.canJump([1, 1, 0]) == True)

# 5. نطّة كبيرة فوق الأصفار
print(s.canJump([4, 0, 0, 0, 1]) == True)

# 6. مقارنة بالحل البطيء
def slow(nums):
    reach = [False] * len(nums)
    reach[0] = True
    for i in range(len(nums)):
        if reach[i]:
            for j in range(i + 1, min(len(nums), i + nums[i] + 1)):
                reach[j] = True
    return reach[-1]

ok = True
for _ in range(500):
    nums = [random.randint(0, 4) for _ in range(random.randint(1, 20))]
    if s.canJump(nums) != slow(nums):
        ok = False
print(ok)
```
