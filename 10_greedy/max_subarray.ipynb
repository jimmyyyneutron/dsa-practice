# Maximum Subarray (LeetCode 53)

```python
import heapq
import random
from typing import List
```

```python
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        sum = 0
        fsum = float("-inf")
        for i in nums :
            sum += i
            if sum > fsum :
                fsum = sum
            if sum < 0 :
                sum = 0
        return fsum
```

## الاختبارات

```python
s = Solution()

# 1. أمثلة LeetCode
print(s.maxSubArray([-2, 1, -3, 4, -1, 2, 1, -5, 4]) == 6)
print(s.maxSubArray([1]) == 1)
print(s.maxSubArray([5, 4, -1, 7, 8]) == 23)

# 2. كل الأرقام سالبة
print(s.maxSubArray([-3, -1, -2]) == -1)

# 3. رقم سالب في النص بيستاهل نعدّيه
print(s.maxSubArray([4, -1, 2, 1]) == 6)

# 4. رقم سالب كبير بيقطع
print(s.maxSubArray([5, -10, 4]) == 5)

# 5. مقارنة بالحل البطيء
def slow(nums):
    best = float("-inf")
    for a in range(len(nums)):
        total = 0
        for b in range(a, len(nums)):
            total += nums[b]
            best = max(best, total)
    return best

ok = True
for _ in range(500):
    nums = [random.randint(-50, 50) for _ in range(random.randint(1, 30))]
    if s.maxSubArray(nums) != slow(nums):
        ok = False
print(ok)
```
