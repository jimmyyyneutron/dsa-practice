# Partition Labels (LeetCode 763)

```python
import heapq
import random
from typing import List
```

```python
class Solution:
    def partitionLabels(self, s: str) -> List[int]:
        lastindex = {}
        for i , c in enumerate(s):
            lastindex[c]=i
        res=[]
        end , size =  0 , 0
        for i , c in enumerate(s):
            size += 1
            if lastindex[c] > end :
                end = lastindex[c]
            if i == end :
                res.append(size)
                size = 0
        return res
```

## الاختبارات

```python
sol = Solution()

# 1. أمثلة LeetCode
print(sol.partitionLabels("ababcbacadefegdehijhklij") == [9, 7, 8])
print(sol.partitionLabels("eccbbbbdec") == [10])

# 2. المثال اللي اشتغلنا عليه
print(sol.partitionLabels("abacd") == [3, 1, 1])

# 3. حرف واحد
print(sol.partitionLabels("a") == [1])

# 4. كل الحروف مختلفة
print(sol.partitionLabels("abc") == [1, 1, 1])

# 5. نفس الحرف متكرر
print(sol.partitionLabels("aaaa") == [4])

# 6. تأكيد الشروط على كلام عشوائي
def valid(s, parts):
    if sum(parts) != len(s):
        return False
    pieces, start = [], 0
    for p in parts:
        pieces.append(set(s[start:start + p]))
        start += p
    for a in range(len(pieces)):
        for b in range(a + 1, len(pieces)):
            if pieces[a] & pieces[b]:
                return False
    return True

ok = True
for _ in range(300):
    s = "".join(random.choice("abcdef") for _ in range(random.randint(1, 25)))
    if not valid(s, sol.partitionLabels(s)):
        ok = False
print(ok)
```
