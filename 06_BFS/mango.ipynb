```python
from collections import deque
```

```python
def main(graph , jobs) :
    search_queue = deque()
    search_queue += graph["you"]
    searched = set()
    while search_queue :
        person = search_queue.popleft()
        if person in searched:
            continue
        if person_is_seller(person , jobs):
            return person
        else :
            searched.add(person)
            search_queue += graph[person]
    return False


def person_is_seller(name, jobs: dict):
    if jobs[name] == "Mango":
        return True
    else :
        return False
    
```

```python
# Test 1: مثال الكتاب، بيّاعين على نفس البعد، فبيرجع اللي دخل الطابور الأول
graph1 = {"you": ["alice", "bob", "claire"], "alice": ["peggy"], "bob": ["anuj", "peggy"],
          "claire": ["thom", "jonny"], "peggy": [], "anuj": [], "thom": [], "jonny": []}
jobs1 = {"alice": "Teacher", "bob": "Engineer", "claire": "Doctor", "peggy": "Mango",
         "anuj": "Driver", "thom": "Mango", "jonny": "Chef"}
assert main(graph1, jobs1) == "peggy"

# Test 2: صاحب مباشر بيّاع، فبيرجعه هو، لأنه الأقرب
jobs2 = dict(jobs1, alice="Mango")
assert main(graph1, jobs2) == "alice"

# Test 3: مفيش بيّاع خالص
jobs3 = {name: "Teacher" for name in jobs1}
assert main(graph1, jobs3) == False

# Test 4: دايرة (a صاحب b، و b صاحب a)، فلازم ميلفّش للأبد
graph4 = {"you": ["a"], "a": ["b"], "b": ["a", "c"], "c": []}
jobs4 = {"a": "Teacher", "b": "Doctor", "c": "Mango"}
assert main(graph4, jobs4) == "c"

# Test 5: مالكش صحاب
assert main({"you": []}, {}) == False

print("BFS passed ✅")
```
