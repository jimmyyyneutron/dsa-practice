```python
import heapq
from typing import List
```

```python
class KthLargest:

    def __init__(self, k: int, nums: List[int]):
        self.nums = nums
        self.k = k
        
        heapq.heapify(nums)
        while len(self.nums) > k:
            heapq.heappop(self.nums)

    def add(self, val: int) -> int:
        heapq.heappush(self.nums , val)
        if len(self.nums) > self.k:
            heapq.heappop(self.nums)
        return self.nums[0]
```

```python
import random

# 1. مثال LeetCode
kl = KthLargest(3, [4, 5, 8, 2])
print(kl.add(3) == 4)
print(kl.add(5) == 5)
print(kl.add(10) == 5)
print(kl.add(9) == 8)
print(kl.add(4) == 8)

# 2. nums فاضية في الأول
kl = KthLargest(1, [])
print(kl.add(-3) == -3)
print(kl.add(-2) == -2)
print(kl.add(-4) == -2)

# 3. nums أقصر من k بعنصر
kl = KthLargest(2, [0])
print(kl.add(-1) == -1)
print(kl.add(1) == 0)

# 4. تكرار
kl = KthLargest(2, [5, 5, 5])
print(kl.add(5) == 5)
print(kl.add(1) == 5)

# 5. أرقام سالبة
kl = KthLargest(2, [-10, -20, -30])
print(kl.add(-5) == -10)
print(kl.add(-1) == -5)

# 6. مقارنة بالحل البطيء على أرقام عشوائية
ok = True
for _ in range(200):
    k = random.randint(1, 10)
    nums = [random.randint(-100, 100) for _ in range(random.randint(k - 1, 30))]
    kl = KthLargest(k, list(nums))
    all_nums = list(nums)
    for _ in range(20):
        val = random.randint(-100, 100)
        all_nums.append(val)
        if kl.add(val) != sorted(all_nums, reverse=True)[k - 1]:
            ok = False
print(ok)
```
