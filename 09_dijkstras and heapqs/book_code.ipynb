```python
# graph
graph = {}
graph["start"] = {}
graph["a"] = {}
graph["b"] = {}
graph["fin"] = {}
graph["start"]["a"] = 6
graph["start"]["b"] = 2
graph["a"]["fin"] = 1
graph["b"]["a"] = 3
graph["b"]["fin"]=5
# cost
cost = {} 
cost["a"] = 6
cost["b"] = 2
cost["fin"] = float("inf")
# parents 
parents = {}
parents["a"] = "start"
parents["b"] = "start"
parents["fin"]= None
processed = []
def cheapest_node(cost):
    lowest_cost = float("inf")
    lowest_node = None
    for i in cost :
        value = cost[i]
        if value < lowest_cost and i not in processed:
            lowest_cost = value 
            lowest_node = i 
    return lowest_node

node = cheapest_node(cost)
while node is not None :
    value = cost[node]
    neighbours = graph[node]
    for n in neighbours.keys() :
        new_cost = value + neighbours[n]
        if cost[n] > new_cost :
            cost[n] = new_cost 
            parents[n] = node 
    processed.append(node)
    node = cheapest_node(cost)
    print (cost)
```
