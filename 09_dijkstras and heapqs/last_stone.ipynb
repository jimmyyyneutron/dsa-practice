```python
from typing import List
import heapq
```

```python
class Solution:
    def lastStoneWeight(self, stones: List[int]) -> int:
        greatest = [-x for x in stones]
        heapq.heapify(greatest)
        while len(greatest) > 1 :
            first = -heapq.heappop(greatest)
            second= -heapq.heappop(greatest)
            if first != second :
                if first < second :
                    first = 0
                    second = second - first 
                    heapq.heappush(greatest , -second)
                elif first > second :
                    first = first - second
                    second = 0
                    heapq.heappush(greatest , -first)
            if not greatest :
                return 0
        return -greatest[0]
    #time:  O(n log n)
    #space: O(n)
```

```python
import random

s = Solution()

# 1. مثال LeetCode
print(s.lastStoneWeight([2, 7, 4, 1, 8, 1]) == 1)

# 2. حجر واحد بس
print(s.lastStoneWeight([1]) == 1)
print(s.lastStoneWeight([5]) == 5)

# 3. حجرين متساويين، فالـ heap يفضى
print(s.lastStoneWeight([3, 3]) == 0)

# 4. حجرين مختلفين
print(s.lastStoneWeight([10, 4]) == 6)

# 5. كله متساوي
print(s.lastStoneWeight([5, 5, 5, 5]) == 0)
print(s.lastStoneWeight([5, 5, 5]) == 5)

# 6. أرقام مترتبة
print(s.lastStoneWeight([1, 2, 3, 4, 5]) == 1)


# 7. مقارنة بالحل البطيء على أرقام عشوائية
def slow(stones):
    stones = list(stones)
    while len(stones) > 1:
        stones.sort()
        a = stones.pop()
        b = stones.pop()
        if a != b:
            stones.append(a - b)
    return stones[0] if stones else 0

ok = True
for _ in range(500):
    stones = [random.randint(1, 1000) for _ in range(random.randint(1, 30))]
    if s.lastStoneWeight(list(stones)) != slow(stones):
        ok = False
print(ok)
```
