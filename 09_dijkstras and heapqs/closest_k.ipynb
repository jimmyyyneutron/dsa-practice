```python
import heapq 
from typing import List
```

```python
class Solution:
    def kClosest(self, points: List[List[int]], k: int) -> List[List[int]]:
        heap = []
        for x , y in points :
            dist = x**2 + y**2 
            point = [x , y]
            heapq.heappush(heap,(dist , point))
        result = []
        while k>0 :
            dist , point = heapq.heappop(heap)
            result.append(point)
            k-=1
        return result
    # time:  O(n log n)
    # space: O(n)
```

```python
import random

s = Solution()

# المسألة بتقبل أي ترتيب، فبنرتّب قبل المقارنة
def same(a, b):
    return sorted(a) == sorted(b)

# 1. أمثلة LeetCode
print(same(s.kClosest([[1, 3], [-2, 2]], 1), [[-2, 2]]))
print(same(s.kClosest([[3, 3], [5, -1], [-2, 4]], 2), [[3, 3], [-2, 4]]))

# 2. نقطة واحدة
print(same(s.kClosest([[0, 1]], 1), [[0, 1]]))

# 3. k = عدد النقط كله
print(same(s.kClosest([[1, 1], [2, 2], [3, 3]], 3), [[1, 1], [2, 2], [3, 3]]))

# 4. نقطة الأصل نفسها
print(same(s.kClosest([[5, 5], [0, 0], [1, 2]], 1), [[0, 0]]))

# 5. أرقام سالبة
print(same(s.kClosest([[-1, -1], [-5, -5], [2, -3]], 2), [[-1, -1], [2, -3]]))

# 6. نقط مكررة
print(same(s.kClosest([[1, 1], [1, 1], [9, 9]], 2), [[1, 1], [1, 1]]))


# 7. مقارنة بالحل البطيء على أرقام عشوائية
def dist(p):
    return p[0] ** 2 + p[1] ** 2

ok = True
for _ in range(300):
    pts = [[random.randint(-50, 50), random.randint(-50, 50)] for _ in range(random.randint(1, 30))]
    k = random.randint(1, len(pts))
    mine = s.kClosest([p[:] for p in pts], k)
    slow = sorted(pts, key=dist)[:k]
    # لو فيه نقط على نفس المسافة، أي اختيار صح، فبنقارن المسافات
    if len(mine) != k or sorted(map(dist, mine)) != sorted(map(dist, slow)):
        ok = False
print(ok)
```
